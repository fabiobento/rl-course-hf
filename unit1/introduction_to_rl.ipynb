# Introdução ao Aprendizado por Reforço

## O que é Aprendizado por Reforço?

Para entender o Aprendizado por Reforço (Reinforcement Learning), vamos começar com a visão geral.

### A visão geral

A ideia por trás do Aprendizado por Reforço(*Reinforcement Learning* — RL) é que um agente (uma IA) aprenderá com o ambiente ao **interagir** com ele (por tentativa e erro) e receber **recompensas** (negativas ou positivas) como feedback por realizar ações.

Aprender a partir das interações com o ambiente **vem das nossas experiências naturais**.


Por exemplo, imagine colocar alguém em frente a um videogame que ele nunca jogou, entregar um controle e deixá-lo sozinho.

<img src="https://github.com/fabiobento/rl-course-hf/raw/main/unit1/images/Illustration_1.jpg" alt="The RL process" width="100%">

Fonte: [Hugging Face Deep Reinforcement Learning Course](https://huggingface.co/learn/deep-rl-course/unit0/introduction?fw=pt)

A pessoa irá interagir com o ambiente (o videogame) pressionando o direcional pra direita (ação).

Ela pegou uma moeda, isso é uma recompensa de +1.
- É positivo, ela acabou de entender que neste jogo **deve pegar as moedas**.


<img src="https://github.com/fabiobento/rl-course-hf/raw/main/unit1/images/Illustration_2.jpg" alt="The RL process" width="100%">

Fonte: [Hugging Face Deep Reinforcement Learning Course](https://huggingface.co/learn/deep-rl-course/unit0/introduction?fw=pt)

Mas então, **ele aperta o direcional pra direita novamente** e encosta em um inimigo.
- Ele acabou de morrer, então isso é uma recompensa de -1.


<img src="https://github.com/fabiobento/rl-course-hf/raw/main/unit1/images/Illustration_3.jpg" alt="The RL process" width="100%">

Fonte: [Hugging Face Deep Reinforcement Learning Course](https://huggingface.co/learn/deep-rl-course/unit0/introduction?fw=pt)

Ao interagir com o ambiente por tentativa e erro, o jogador entende que:
- **precisa pegar as moedas nesse ambiente, mas evitar os inimigos.**


É assim que humanos e animais aprendem, **por meio da interação**.

> O Aprendizado por Reforço é apenas uma **abordagem computacional de aprender a partir das ações**.

### Uma definição mais formal

Podemos agora fazer uma definição mais formal do Aprendizado por Reforço:

> O aprendizado por reforço é um *framework* para resolver tarefas de controle (também chamadas de problemas de decisão) por meio da criação de agentes que aprendem com o ambiente interagindo com ele por meio de tentativa e erro e recebendo recompensas (positivas ou negativas) como feedback exclusivo.

Mas, como o Aprendizado por Reforço Funciona?

## O Framework do Aprendizado por Reforço

### O processo RL

O processo RL: um loop de estado, ação, recompensa e próximo estado.
<img src="https://github.com/fabiobento/rl-course-hf/raw/main/unit1/images/RL_process.jpg" alt="The RL process" width="100%">

Fonte: [Reinforcement Learning: An Introduction, Richard Sutton and Andrew G. Barto](http://incompleteideas.net/book/RLbook2020.pdf)

Para entender o processo RL, vamos imaginar um agente aprendendo a jogar um jogo de plataforma:

<img src="https://github.com/fabiobento/rl-course-hf/raw/main/unit1/images/RL_process_game.jpg" alt="The RL process" width="100%">

Fonte: [Hugging Face Deep Reinforcement Learning Course](https://huggingface.co/learn/deep-rl-course/unit0/introduction?fw=pt)

- Nosso Agente recebe o **`state`** $S_0$ do **`Environment`** — recebemos o primeiro frame de nosso jogo (o `Environment`).
- Com base no **`state`** $S_0$, o agente executa a **`action`** $A_0$ — nosso agente se move para a direita.
- O ambiente vai a para um novo **`state`** $S_1$ — novo frame.
- O ambiente dá uma **reward** $R_1$ para o agente — não estamos mortos (*Positive Reward* $+1$).

Esse loop retorna uma sequência `state`, `action`, `reward`, `next state`.
<img src="https://github.com/fabiobento/rl-course-hf/raw/main/unit1/images/sars.jpg" alt="The RL process" width="100%">

Fonte: [Hugging Face Deep Reinforcement Learning Course](https://huggingface.co/learn/deep-rl-course/unit0/introduction?fw=pt)

O objetivo do agente é **maximizar a recompensa acumulativa**, chamada de retorno esperado(**_expected return_**).

### A hipótese da recompensa: a ideia central do Aprendizado por Reforço

> Por que o objetivo do agente é maximizar o retorno esperado?

Porque a RL se baseia na **hipótese da recompensa**, ou seja, todas as metas podem ser descritas como a **maximização do retorno esperado** (recompensa cumulativa esperada).

É por isso que, no Aprendizado por Reforço, **para ter o melhor comportamento**, nosso objetivo é aprender a realizar ações que **maximizem a recompensa cumulativa esperada**.

### A propriedade de Markov

Em artigos, você verá que o processo de RL é chamado de **Processo de Decisão de Markov** (*Markov Decision Process* — MDP).

Voltaremos a falar sobre a propriedade de Markov nas próximas unidades.
Mas se você precisa se lembrar de algo hoje sobre isso, é o seguinte:
> a Propriedade de Markov implica que nosso agente precisa **apenas do estado atual para decidir** qual ação tomar e **não do histórico de todos os estados e ações** que ele tomou antes.

### Espaço de Observações/Estados

Observações/Estados são as **informações que nosso agente obtém do ambiente**.

No caso de um videogame, pode ser um frame (uma captura de tela).

No caso de agente de *trading*, pode ser o valor de uma determinada ação, etc.

No entanto, é preciso fazer uma diferenciação entre *observação* e *estado*:

- Estado: é **uma descrição completa do estado do mundo** (não há informações ocultas). Em um ambiente totalmente observado.

No jogo de xadrez, recebemos um estado do ambiente, pois temos acesso a todas as informações do tabuleiro de xadrez.

<img src="https://github.com/fabiobento/rl-course-hf/raw/main/unit1/images/chess.jpg" alt="The RL process" width="100%">

Fonte: [Hugging Face Deep Reinforcement Learning Course](https://huggingface.co/learn/deep-rl-course/unit0/introduction?fw=pt)

Em um jogo de xadrez, temos acesso a todas as informações do tabuleiro, portanto, recebemos um estado do ambiente.

Em outras palavras, o ambiente é totalmente observado.

- Observação: é uma **descrição parcial do estado**. Em um ambiente parcialmente observado.

Em Super Mario Bros, só vemos a parte do nível próximo ao jogador, portanto, recebemos uma observação.

Em Super Mario Bros, estamos em um ambiente parcialmente observado.

### Espaço de ação

Espaço de ação é o conjunto de **todas as possíveis ações em um ambiente**.

As ações podem vir de um espaço *discreto* ou *contínuo*:

- *Espaço discreto*: a quantidade de possíveis ações é **finita**.

Em Super Mario Bros, temos apenas quatro ações possíveis: esquerda, direita, para cima (pular) e para baixo (agachar).
