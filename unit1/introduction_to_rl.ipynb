# Introdução ao Aprendizado por Reforço

## O que é Aprendizado por Reforço?

Para entender o Aprendizado por Reforço (Reinforcement Learning), vamos começar com a visão geral.

### A visão geral

A ideia por trás do Aprendizado por Reforço(*Reinforcement Learning* — RL) é que um agente (uma IA) aprenderá com o ambiente ao **interagir** com ele (por tentativa e erro) e receber **recompensas**(_rewards_) negativas ou positivas como feedback por realizar _actions_.

Aprender a partir das interações com o ambiente **vem das nossas experiências naturais**.


Por exemplo, imagine colocar alguém em frente a um videogame que ele nunca jogou, entregar um controle e deixá-lo sozinho.

<img src="https://github.com/fabiobento/rl-course-hf/raw/main/unit1/images/Illustration_1.jpg" alt="The RL process" width="100%">
<span style="font-size:80%">
Fonte: <a href="https://huggingface.co/learn/deep-rl-course/unit0/introduction?fw=pt" target="_blank">Hugging Face Deep Reinforcement Learning Course</a>
</span>

A pessoa irá interagir com o ambiente (o videogame) pressionando o direcional pra direita (ação).

Ela pegou uma moeda, isso é uma recompensa de +1.
- É positivo, ela acabou de entender que neste jogo **deve pegar as moedas**.


<img src="https://github.com/fabiobento/rl-course-hf/raw/main/unit1/images/Illustration_2.jpg" alt="The RL process" width="100%">
<span style="font-size:80%">
Fonte: <a href="https://huggingface.co/learn/deep-rl-course/unit0/introduction?fw=pt" target="_blank">Hugging Face Deep Reinforcement Learning Course</a>
</span>

Mas então, **ele aperta o direcional pra direita novamente** e encosta em um inimigo.
- Ele acabou de morrer, então isso é uma recompensa de -1.


<img src="https://github.com/fabiobento/rl-course-hf/raw/main/unit1/images/Illustration_3.jpg" alt="The RL process" width="100%">
<span style="font-size:80%">

</span>

Ao interagir com o ambiente por tentativa e erro, o jogador entende que:
- **precisa pegar as moedas nesse ambiente, mas evitar os inimigos.**


É assim que humanos e animais aprendem, **por meio da interação**.

> O Aprendizado por Reforço é apenas uma **abordagem computacional de aprender a partir das _actions_**.

### Uma definição mais formal

Podemos agora fazer uma definição mais formal do Aprendizado por Reforço:

> O aprendizado por reforço é um *framework* para resolver tarefas de controle (também chamadas de problemas de decisão) por meio da criação de agentes que aprendem com o ambiente interagindo com ele por meio de tentativa e erro e recebendo _rewards_ (positivas ou negativas) como feedback exclusivo.

Mas, como o Aprendizado por Reforço Funciona?

- O ambiente dá uma **reward** $R_1$ para o agente — não estamos mortos (*Positive Reward* $+1$).

O objetivo do agente é **maximizar** a recompensa acumulativa(**_cumulative reward_**), chamada de retorno esperado(**_expected return_**).

### A hipótese da recompensa(_reward hypothesis_): a ideia central do Aprendizado por Reforço

> Por que o objetivo do agente é maximizar o retorno esperado?

Porque a RL se baseia na **_reward hypothesis_**, ou seja, todas as metas podem ser descritas como a **maximização do retorno esperado** (_expected cumulative reward_).

É por isso que, no Aprendizado por Reforço, **para ter o melhor comportamento**(_best behavior_), nosso objetivo é aprender a realizar _actions_ que **maximizem a _expected cumulative reward_**.

### A propriedade de Markov

Em artigos, você verá que o processo de RL é chamado de **Processo de Decisão de Markov** (*Markov Decision Process* — MDP).

Voltaremos a falar sobre a propriedade de Markov nas próximas unidades.
Mas se você precisa se lembrar de algo hoje sobre isso, é o seguinte:
> a Propriedade de Markov implica que nosso agente precisa **apenas do _state_ atual para decidir** qual _action_ tomar e **não do histórico de todos os _states_ e _actions_** que ele tomou antes.

No entanto, na realidade, **não podemos simplesmente adicioná-las dessa forma**.

As _rewards_ que chegam mais cedo (no início do jogo) **têm mais probabilidade de acontecer**, pois são mais previsíveis do que a _reward_ futura de longo prazo.

Vamos supor que seu agente é esse rato que pode se mover um bloco por vez, e seu oponente é o gato (que também pode se mover).

O objetivo do rato é **comer a maior quantidade de queijo possível antes de ser pego pelo gato**.

<img src="https://github.com/fabiobento/rl-course-hf/raw/main/unit1/images/rewards_3.jpg" alt="The RL process" width="100%">
<span style="font-size:80%">

</span>

Como podemos ver no diagrama, **é mais provável comer o queijo que está perto de nós do que o queijo próximo ao gato** (quanto mais perto estamos do gato, mais perigoso é).

Consequentemente, **a _reward_ próxima ao gato, mesmo que seja maior (mais queijo), será mais descontada**, já que não temos certeza se conseguiremos comê-la.

Para descontar as _rewards_, fazemos assim:
1. Definimos uma taxa de desconto chamada gamma($\gamma$). **Ela deve estar entre $0$ e $1$**. Na maioria das vezes, entre **$0,95$ e $0,99$**.
- Quanto maior o $\gamma$, menor o desconto. Isso significa que nosso agente **se importa mais com a _reward_ de longo prazo**.
- Por outro lado, quanto menor o $\gamma$, maior o desconto. Isso significa que nosso **agente se importa mais com a _reward_ de curto prazo (o queijo mais próximo)**.

2. Então, cada _reward_ será descontada por $\gamma$ elevado ao expoente do passo de tempo. À medida que o passo de tempo aumenta, o gato se aproxima de nós, então a _reward_ futura é cada vez menos provável de acontecer.

## Tipos de tarefas

Uma tarefa é uma **instância** de um problema de RL.

Podemos ter dois tipos de tarefas:
- **episódicas** e,
- **contínuas**

### Tarefa Episódica

Neste caso, temos um ponto de início e um ponto de término **(um _state_ terminal)**. 

**Isso cria um episódio**: uma lista de _States_, _Actions_, _Rewards_ e novos _States_.

Por exemplo, pense no Super Mario Bros: um episódio começa ao iniciar um novo nível do Mario e termina **quando você é derrotado ou chega ao final do nível**.

## O trade-off de Exploration/Exploitation

Por fim, antes de analisar os diferentes métodos para resolver problemas de RL, precisamos abordar mais um tópico muito importante: o *trade-off de exploration/exploitation*.


- *Exploration* é explorar o ambiente tentando _actions_ aleatórias para **encontrar mais informações sobre o ambiente**.
- *Explotation* é o **aproveitamento de informações conhecidas para maximizar a _reward_**.

Lembre-se de que o objetivo do nosso agente de RL é maximizar a _reward_ cumulativa esperada.

No entanto, **podemos cair em uma armadilha comum**.

Vamos ver um exemplo:

<img src="https://github.com/fabiobento/rl-course-hf/raw/main/unit1/images/exp_1.jpg" alt="The RL process" width="100%">

<span style="font-size:80%">
Fonte: <a href="https://huggingface.co/learn/deep-rl-course/unit0/introduction?fw=pt" target="_blank">Hugging Face Deep Reinforcement Learning Course</a>
</span>

Nesse jogo, nosso camundongo pode ter uma **quantidade infinita de queijos pequenos** (+1 cada). Mas, no topo do labirinto, há uma soma gigantesca de queijo (+1000).

No entanto, se nos concentrarmos apenas em *explotation*, nosso agente nunca alcançará a gigantesca soma de queijo.

Em vez disso, ele só aproveitará **a fonte de _rewards_ mais próxima**, mesmo que essa fonte seja pequena (*exploitation*).

Mas se nosso agente explorar um pouco, ele poderá **descobrir a grande _reward_** (a pilha de queijo grande).

Isso é o que chamamos de trade-off exploration/explotation. Precisamos equilibrar o quanto **exploramos o ambiente** e o quanto **aproveitamos o que sabemos sobre ele**.

Portanto, precisamos **definir uma regra que ajude a lidar com esse trade-off**. Veremos as diferentes maneiras de lidar com isso nas próximas unidades.

## As duas principais abordagens para resolver problemas de RL

> Agora que aprendemos o framework, como resolvemos o problema de RL?

Em outras palavras, como criamos um agente de RL que pode **selecionar as _actions_ que maximizam sua _expected cumulative reward_**?

### A _Policy_ $\pi$: o cérebro do agente

A _Policy_ $\pi$ é o **cérebro do nosso Agente**, é a função que nos diz **qual _action_ executar dado o estado em que nos encontramos**. Portanto, ela define o **comportamento do agente** em um determinado momento.

<img src="https://github.com/fabiobento/rl-course-hf/raw/main/unit1/images/policy_1.jpg" alt="The RL process" width="100%">

<span style="font-size:80%">
Pense na _policy_ como o cérebro do nosso agente, a função que nos dirá a _action_ a ser tomada em um determinado _state_

Fonte: <a href="https://huggingface.co/learn/deep-rl-course/unit0/introduction?fw=pt" target="_blank">Hugging Face Deep Reinforcement Learning Course</a>
</span>

Essa _policy_ **é a função que queremos aprender**:
- Nosso objetivo é encontrar a _policy_ ideal $\pi$*, a _policy_ que **maximiza o retorno esperado** quando o agente age de acordo com ela.
- Encontramos essa $\pi$* **por meio de treinamento**.

Há duas abordagens para treinar nosso agente a encontrar essa _policy_ ideal $\pi$*:
- **Diretamente**, ensinando o agente a aprender **qual _action_ executar**, dado o _state_ atual: (_Policy-Based Methods_).
- Indiretamente, **ensinar o agente a aprender qual _state_ é mais valioso** e, em seguida, executar a _action_ que **leva aos _states_ mais valiosos**: **Métodos baseados em valor**(_Value-Based Methods_).

#### _Policy-Based Methods_

Em métodos baseados em _policy_ , aprendemos uma _policy function_ diretamente.

Essa função definirá um mapeamento de cada _state_ para a melhor _action_ correspondente.

Uma outra opção é ela  definir uma **distribuição de probabilidade sobre o conjunto de possíveis _actions_ naquele _state_**.
